```python
import os, sys
# Add src folder to import path
sys.path.append("../src")
from loader import load_documents
from rag_pipeline import build_vector_db, create_qa_chain
from evaluation import create_bloom_evaluator, format_evaluation_output
from visualization import visualize_vector_space
```

```python
# 🔑 IMPORTANT:
# Add your Hugging Face token manually before running this notebook:
os.environ["HUGGINGFACEHUB_API_TOKEN"] = "YOUR_TOKEN_HERE"
```

```python
docs = load_documents("../data")
```

```python
vector_db = build_vector_db(docs)
```

```python
qa = create_qa_chain(vector_db)
```

```python
question = "what is data dignity?"
response = qa.invoke({"question": question})
print(response)
```

```python
# Create evaluator
bloom_eval = create_bloom_evaluator()

# Get context
context_docs = qa.retriever.invoke(question)
context = " ".join([doc.page_content for doc in context_docs[:2]])

# Answer from model
answer = qa.invoke({"question": question})["result"]

# Evaluate
evaluation_input = {"context": context, "answer": answer}
result = bloom_eval.invoke(evaluation_input)

print("🧩 Question:", question)
print("💬 LLM Answer:", answer)
print(format_evaluation_output(question, answer, result))
```
